```python
f=open('./dic.utf8','r',encoding='utf-8-sig')
```

```python
lines=f.readlines()
```

```python
lines
```

```python
f.close()
```

```python
dic=[line.strip() for line in lines]
```

```python
dic
```

```python
text='北京市民办高中'
```

```python
len(text)
```

```python
max_len=0
for word in dic:
    if len(word)>max_len:
        max_len=len(word)
```

```python
max_len
```

```python
tl=len(text)
```

```python
start=0
```

```python
result=[ ]
```

```python
start=0
while start<tl:
    end=start+max_len
for _ in range(max_len):
        t=text[start:end]
        if (t in dic)or(len(t)==1):
            result.append(t)
            start=end
            break
        end-=1
```
