```python
import pandas as pd
import numpy as np
from pyecharts import options as opts
from pyecharts.charts import Bar, Line, Pie, Grid, Page
from pyecharts.components import Table
from pyecharts.globals import ThemeType
import re
from pyecharts.commons.utils import JsCode
import os


# 定义文件路径
file_path = './中国电影网电影_data (1).xlsx'

def load_data(file_path):
    """加载Excel数据并进行全面错误处理"""
    if not os.path.exists(file_path):
        print(f"错误：文件 {file_path} 不存在，请检查文件路径是否正确。")
        print("提示：请确保文件与代码在同一目录下，或提供完整路径。")
        return None
    
    try:
        df = pd.read_excel(file_path)
        print(f"成功加载数据，共 {len(df)} 条记录。")
        return df
    except Exception as e:
        print(f"读取文件时发生错误：{e}")
        return None

# 加载数据
df = load_data(file_path)
if df is None:
    exit()  # 数据加载失败时退出程序

# 数据预处理函数
def parse_box_office(box_office):
    """处理票房数据，提取数字并转换为亿元"""
    if isinstance(box_office, str):
        match = re.search(r'(\d+\.\d+|\d+)', box_office)
        if match:
            return float(match.group(1)) / 10000  # 万转亿
    return 0

def parse_country(country):
    """简化国家及地区名称"""
    if isinstance(country, str):
        if '中国' in country:
            return '中国'
        elif '美国' in country:
            return '美国'
        elif '香港' in country:
            return '中国香港'
        elif '台湾' in country:
            return '中国台湾'
        elif '日本' in country:
            return '日本'
        elif '韩国' in country:
            return '韩国'
        else:
            return '其他'
    return '其他'

def parse_first_genre(genre):
    """提取第一个电影类型"""
    if isinstance(genre, str):
        return genre.split('/')[0]
    return '未知'

# 执行数据预处理
df['累计票房(亿元)'] = df['累计票房'].apply(parse_box_office)
df['上映年份'] = pd.to_datetime(df['上映时间']).dt.year.fillna(0).astype(int)
df['国家简化'] = df['国家及地区'].apply(parse_country)
df['主要类型'] = df['类型'].apply(parse_first_genre)
df['是否国产'] = df['国家及地区'].apply(lambda x: 1 if '中国' in str(x) and not ('美国' in str(x) or '日本' in str(x) or '韩国' in str(x)) else 0)

# 1. 业绩总览图表
total_box_office = df['累计票房(亿元)'].sum()

yearly_box_office = df.groupby('上映年份')['累计票房(亿元)'].sum()
total_yearly = yearly_box_office.sum()
yearly_percentage = (yearly_box_office / total_yearly * 100).round(2).reset_index(name='占比(%)')

# 筛选2016-2018年数据
recent_years = yearly_percentage[yearly_percentage['上映年份'].isin([2016, 2017, 2018])]
other_years = yearly_percentage[~yearly_percentage['上映年份'].isin([2016, 2017, 2018])]
other_percent = other_years['占比(%)'].sum()
recent_years = recent_years.append(pd.DataFrame({
    '上映年份': ['其他年份', '累计票房(亿元)'],
    '占比(%)': [other_percent, total_box_office]
}))

# 创建业绩总览表格（完整闭合）
title_table = Table()
title_table.add(
    headers=["指标", "数值"],
    rows=[
        ["累计票房收入(亿元)", f"{total_box_office:.2f}"],
        ["2018年度占比", f"{recent_years[recent_years['上映年份'] == 2018]['占比(%)'].values[0] if 2018 in recent_years['上映年份'].values else 0}%"],
        ["2017年度占比", f"{recent_years[recent_years['上映年份'] == 2017]['占比(%)'].values[0] if 2017 in recent_years['上映年份'].values else 0}%"],
        ["2016年度占比", f"{recent_years[recent_years['上映年份'] == 2016]['占比(%)'].values[0] if 2016 in recent_years['上映年份'].values else 0}%"],
    ],
)
title_table.set_global_opts(
    title_opts=opts.TitleOpts(title="业绩总览", subtitle="中国电影票房关键指标"),
    toolbox_opts=opts.ToolboxOpts(is_show=False),
    table_opts=opts.TableOpts(
        cell_style=opts.CellStyleOpts(
            color=["#fff", "#f5f5f5"],
            font_size=16,
            font_weight="bold"
        ),
        head_style=opts.CellStyleOpts(
            color="#333",
            bg_color="#409EFF",
            font_size=18,
            font_weight="bold"
        )
    )
)

# 2. 票房前十影片排行榜
top10_movies = df.nlargest(10, '累计票房(亿元)', keep='all')

bar_top10 = (
    Bar(init_opts=opts.InitOpts(width="100%", height="400px", theme=ThemeType.LIGHT))
    .add_xaxis(top10_movies['电影名'].tolist())
    .add_yaxis("累计票房(亿元)", top10_movies['累计票房(亿元)'].tolist(), color="#5470c6")
    .set_global_opts(
        title_opts=opts.TitleOpts(title="票房前十影片排行榜"),
        xaxis_opts=opts.AxisOpts(axislabel_opts={"rotate": 45}),
        yaxis_opts=opts.AxisOpts(name="票房(亿元)"),
        toolbox_opts=opts.ToolboxOpts(is_show=True),
        legend_opts=opts.LegendOpts(is_show=False)
    )
    .set_series_opts(
        label_opts=opts.LabelOpts(position="top")
    )
)

# 3. 各个国家影片数量柱状图
country_movie_count = df['国家简化'].value_counts().reset_index()
country_movie_count.columns = ['国家', '影片数量']

bar_country = (
    Bar(init_opts=opts.InitOpts(width="100%", height="400px", theme=ThemeType.LIGHT))
    .add_xaxis(country_movie_count['国家'].tolist())
    .add_yaxis("影片数量", country_movie_count['影片数量'].tolist(), color="#91cc75")
    .set_global_opts(
        title_opts=opts.TitleOpts(title="各国影片数量对比"),
        yaxis_opts=opts.AxisOpts(name="影片数量"),
        toolbox_opts=opts.ToolboxOpts(is_show=True),
        legend_opts=opts.LegendOpts(is_show=False)
    )
    .set_series_opts(
        label_opts=opts.LabelOpts(position="top")
    )
)

# 4. 年度票房总额和影片数量（修改列名避免括号）
yearly_stats = df.groupby('上映年份').agg(
    票房总额亿元=('累计票房(亿元)', 'sum'),  # 修改为合法变量名
    影片数量=('电影名', 'count')
).reset_index()

bar_line_year = (
    Bar(init_opts=opts.InitOpts(width="100%", height="400px", theme=ThemeType.LIGHT))
    .add_xaxis(yearly_stats['上映年份'].tolist())
    .add_yaxis("票房总额(亿元)", yearly_stats['票房总额亿元'].tolist(), color="#5470c6")  # 使用新列名
    .extend_axis(
        yaxis=opts.AxisOpts(
            name="影片数量",
            type_="value",
            min_=0,
            max_=yearly_stats['影片数量'].max() * 1.1,
            interval=yearly_stats['影片数量'].max() // 5 if yearly_stats['影片数量'].max() > 0 else 1,
            axislabel_opts=opts.LabelOpts(formatter="{value} 部")
        )
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="各年度票房总额和影片数量"),
        xaxis_opts=opts.AxisOpts(name="年份"),
        yaxis_opts=opts.AxisOpts(name="票房(亿元)"),
        toolbox_opts=opts.ToolboxOpts(is_show=True),
        legend_opts=opts.LegendOpts(pos_top="90%")
    )
)

line_year = (
    Line()
    .add_xaxis(yearly_stats['上映年份'].tolist())
    .add_yaxis(
        "影片数量",
        yearly_stats['影片数量'].tolist(),
        yaxis_index=1,
        color="#d14a61",
    )
)

bar_line_year.overlap(line_year)

# 5. 电影公司影片数量和票房总额（修改列名避免括号）
company_stats = df.groupby('发行公司').agg(
    影片数量=('电影名', 'count'),
    票房总额亿元=('累计票房(亿元)', 'sum')  # 修改为合法变量名
).reset_index()

top20_companies = company_stats.nlargest(20, '影片数量')

line_company = (
    Line(init_opts=opts.InitOpts(width="100%", height="400px", theme=ThemeType.LIGHT))
    .add_xaxis(top20_companies['发行公司'].tolist())
    .add_yaxis(
        "影片数量",
        top20_companies['影片数量'].tolist(),
        yaxis_index=0,
        color="#5470c6",
        label_opts=opts.LabelOpts(is_show=False)
    )
    .extend_axis(
        yaxis=opts.AxisOpts(
            name="票房总额(亿元)",
            type_="value",
            min_=0,
            max_=top20_companies['票房总额亿元'].max() * 1.1,  # 使用新列名
            axislabel_opts=opts.LabelOpts(formatter="{value} 亿")
        )
    )
    .add_yaxis(
        "票房总额(亿元)",
        top20_companies['票房总额亿元'].tolist(),  # 使用新列名
        yaxis_index=1,
        color="#91cc75",
        label_opts=opts.LabelOpts(is_show=False)
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="主要电影公司影片数量和票房"),
        xaxis_opts=opts.AxisOpts(axislabel_opts={"rotate": 45}),
        yaxis_opts=[
            opts.AxisOpts(name="影片数量"),
            opts.AxisOpts(name="票房总额(亿元)")
        ],
        toolbox_opts=opts.ToolboxOpts(is_show=True),
        legend_opts=opts.LegendOpts(pos_top="90%")
    )
)

# 6. 纯国产电影年度走势（修改列名避免括号）
chinese_movies = df[df['是否国产'] == 1]
chinese_yearly = chinese_movies.groupby('上映年份').agg(
    票房总额亿元=('累计票房(亿元)', 'sum'),  # 修改为合法变量名
    影片数量=('电影名', 'count')
).reset_index()

line_chinese = (
    Line(init_opts=opts.InitOpts(width="100%", height="400px", theme=ThemeType.LIGHT))
    .add_xaxis(chinese_yearly['上映年份'].tolist())
    .add_yaxis(
        "票房总额(亿元)",
        chinese_yearly['票房总额亿元'].tolist(),  # 使用新列名
        symbol="circle",
        color="#5470c6",
        label_opts=opts.LabelOpts(is_show=False)
    )
    .add_yaxis(
        "影片数量",
        chinese_yearly['影片数量'].tolist(),
        yaxis_index=1,
        symbol="triangle",
        color="#d14a61",
        label_opts=opts.LabelOpts(is_show=False)
    )
    .extend_axis(
        yaxis=opts.AxisOpts(
            name="影片数量",
            type_="value",
            min_=0,
            max_=chinese_yearly['影片数量'].max() * 1.1,
            axislabel_opts=opts.LabelOpts(formatter="{value} 部")
        )
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="纯国产电影年度走势"),
        xaxis_opts=opts.AxisOpts(name="年份"),
        yaxis_opts=[
            opts.AxisOpts(name="票房总额(亿元)"),
            opts.AxisOpts(name="影片数量")
        ],
        toolbox_opts=opts.ToolboxOpts(is_show=True),
        legend_opts=opts.LegendOpts(pos_top="90%")
    )
)

# 7. 年度类别票房数据
genre_yearly = df.groupby(['上映年份', '主要类型'])['累计票房(亿元)'].sum().reset_index()

table_genre_yearly = (
    Table(init_opts=opts.InitOpts(width="100%", height="400px", theme=ThemeType.LIGHT))
    .add(
        headers=["年份", "类型", "票房总额(亿元)"],
        rows=genre_yearly[['上映年份', '主要类型', '累计票房(亿元)']]
            .sort_values(by=['上映年份', '累计票房(亿元)', '主要类型'], ascending=[True, False, True])
            .round(2)
            .values.tolist(),
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="各年度各类型票房分布"),
        toolbox_opts=opts.ToolboxOpts(is_show=True),
        table_opts=opts.TableOpts(
            cell_style=opts.CellStyleOpts(
                color=["#fff", "#f5f5f5"],
                font_size=14
            ),
            head_style=opts.CellStyleOpts(
                color="#333",
                bg_color="#409EFF",
                font_size=16,
                font_weight="bold"
            )
        )
    )
)

# 8. 类别数量饼图对比
genre_count = df['主要类型'].value_counts().reset_index()
genre_count.columns = ['类型', '影片数量']

pie_genre = (
    Pie(init_opts=opts.InitOpts(width="100%", height="400px", theme=ThemeType.LIGHT))
    .add(
        "影片类型分布",
        [list(z) for z in zip(genre_count['类型'], genre_count['影片数量'])],
        radius=["30%", "75%"],
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="各类型影片数量占比"),
        toolbox_opts=opts.ToolboxOpts(is_show=True, orient="vertical", pos_top="15%", pos_left="2%"),
        legend_opts=opts.LegendOpts(type_="scroll", pos_top="90%", pos_left="center")
    )
    .set_series_opts(
        label_opts=opts.LabelOpts(formatter="{b}: {c} 部 ({d}%)")
    )
)

# 组合所有图表
grid = (
    Grid(init_opts=opts.InitOpts(width="100%", height="1000px", theme=ThemeType.LIGHT))
    .add(title_table, grid_opts=opts.GridOpts(pos_top="5%", pos_left="35%", pos_right="35%"))
    .add(bar_top10, grid_opts=opts.GridOpts(pos_top="20%", pos_left="5%", pos_right="5%", height="20%"))
    .add(bar_country, grid_opts=opts.GridOpts(pos_top="45%", pos_left="5%", pos_right="5%", height="20%"))
    .add(bar_line_year, grid_opts=opts.GridOpts(pos_top="20%", pos_left="35%", pos_right="35%", height="20%"))
    .add(line_company, grid_opts=opts.GridOpts(pos_top="45%", pos_left="35%", pos_right="35%", height="20%"))
    .add(line_chinese, grid_opts=opts.GridOpts(pos_top="20%", pos_left="65%", pos_right="5%", height="20%"))
    .add(table_genre_yearly, grid_opts=opts.GridOpts(pos_top="45%", pos_left="65%", pos_right="5%", height="20%"))
    .add(pie_genre, grid_opts=opts.GridOpts(pos_top="70%", pos_left="35%", pos_right="35%", height="25%"))
)

# 设置全局背景
grid.set_global_opts(
    opts.InitOpts(
        bg_color=JsCode("""
            document.body.style.backgroundColor = '#f0f2f5';
            return '#f0f2f5';
        """)
    )
)

# 保存为HTML文件
grid.render("./电影数据可视化大屏.html")
print("可视化大屏已生成，文件路径：./电影数据可视化大屏.html")
         
```
