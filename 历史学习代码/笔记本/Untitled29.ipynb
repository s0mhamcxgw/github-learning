```python
n=int(input("请输入你要输出的杨辉三角行数："))
list=[]
for i in range(n):
    now=[1]*(i+1)
    for j in range(1,i):
        now[j]=list[i-1][j-1]+list[i-1][j]
    list.append(now)

for line in list:
    print(line)
```
