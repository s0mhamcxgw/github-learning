```python
SOS_token = 0#句子开始标记（Start Of Sentence）
EOS_token = 1
MAX_LENGTH1 = 10#句子的最大长度限制
import jieba
from utils import normalizeString
from utils import cht_to_chs

class Lang:
    def __init__(self, name):
        self.name = name   # 语言名称（如"en"、"zh"）
        self.word2index = {} # 单词 → 索引的映射

        self.word2count = {}# 单词出现次数的统计
        self.index2word = {
            0:"SOS", 1:"EOS"
        }
        self.n_words = 2
        
    def addWord(self, word):#将单词添加到词汇表中，并更新统计信息
        if word not in self.word2index:#如果单词是新词，分配一个新索引（从2开始）。
            self.word2index[word] = self.n_words
            self.word2count[word] = 1
            self.index2word[self.n_words] = word
            self.n_words += 1
        else:
            self.word2count[word] += 1#如果单词已存在，增加其出现次数
            
    def addSentence(self, sentence):#将句子按空格分割后，逐个调用 addWord。
        for word in sentence.split(" "):
            self.addWord(word)
            
def readLangs(lang1, lang2, path):#数据预处理
    lines = open(path, encoding="utf-8").readlines()# 读取文件
    
    lang1_cls = Lang(lang1) # 初始化两种语言的Lang对象
    lang2_cls = Lang(lang2)
    
    pairs = []# 存储有效的句子对
    for l in lines:
        l = l.split("\t") # 假设文件格式为：句子1\t句子2
        sentence1 = normalizeString(l[0])# 规范化句子1（如英文小写化、去标点）
        sentence2 = cht_to_chs(l[1])# 将繁体中文转换为简体中文
        
        seg_list  = jieba.cut(sentence2, cut_all=False)
        sentence2 = " ".join(seg_list)  # 分词后拼接成空格分隔的字符串

        
        if len(sentence1.split(" ")) > MAX_LENGTH1:# 过滤超长句子
            continue
        if len(sentence2.split(" ")) > MAX_LENGTH1:
            continue
            
        pairs.append([sentence1,sentence2])# 保存有效句子对，并更新词汇表
        lang1_cls.addSentence(sentence1)
        lang2_cls.addSentence(sentence2)
        
    return lang1_cls, lang2_cls, pairs
```

```python
# 导入PyTorch深度学习框架的核心模块
import torch       # PyTorch主库，提供张量操作和基础功能
import torch.nn as nn       # 神经网络模块，包含各种层结构（如Linear, LSTM等）和损失函数
import torch.nn.functional as F  # 函数式接口，包含激活函数、损失函数等无状态操作

# 典型用途：
# - torch: 用于张量创建和数学运算（如矩阵乘法）
# - nn: 用于构建神经网络模型（定义模型结构）
# - F: 用于调用ReLU、softmax等函数式操作
```

```python
class EncoderRNN(nn.Module):
    """
    RNN编码器模块，将输入序列编码为隐藏状态表示
    
    Args:
        input_size (int): 输入词汇表大小（不同单词的数量）
        hidden_size (int): 隐藏层维度大小（GRU单元的输出维度）
    """
    def __init__(self, input_size, hidden_size):
        """
        初始化编码器结构
        - input_size: 词汇表大小（如10,000个单词）
        - hidden_size: 隐藏层维度（如256维）
        """
        super(EncoderRNN, self).__init__()  # 调用父类初始化
        self.hidden_size = hidden_size  # 存储隐藏层维度
        
        # 词嵌入层：将单词索引映射为稠密向量
        # 输入形状：[batch_size] → 输出形状：[batch_size, hidden_size]
        self.embedding = nn.Embedding(input_size, hidden_size)
        
        # GRU循环神经网络层
        # 输入输出维度都是hidden_size（保持维度一致）
        self.gru = nn.GRU(hidden_size, hidden_size)
        
    def forward(self, input, hidden):
        """
        前向传播过程
        
        Args:
            input (Tensor): 当前时间步的输入单词索引，形状 [1]
            hidden (Tensor): 上一时间步的隐藏状态，形状 [1, 1, hidden_size]
            
        Returns:
            output (Tensor): 当前时间步的输出，形状 [1, 1, hidden_size]
            hidden (Tensor): 更新后的隐藏状态，形状 [1, 1, hidden_size]
        """
        # 词嵌入 + 维度调整
        # input形状：[1] → embedded形状：[1, 1, hidden_size]
        embedded = self.embedding(input).view(1, 1, -1)
        
        # GRU处理（自动使用上一个隐藏状态）
        # output形状：[1, 1, hidden_size]（seq_len=1, batch=1, hidden_size）
        output, hidden = self.gru(embedded, hidden)
        
        return output, hidden
    
    def initHidden(self):
        """
        初始化全零隐藏状态
        
        Returns:
            Tensor: 形状为 [1, 1, hidden_size] 的全零张量
                   第一个1：num_layers（单层GRU）
                   第二个1：batch_size（单个样本）
        """
        return torch.zeros(1, 1, self.hidden_size, device=device)  # 确保与模型同设备
```

```python
class DecoderRNN(nn.Module):
    """基于GRU的RNN解码器实现，用于序列生成任务"""
    
    def __init__(self, hidden_size, output_size):
        """
        初始化解码器
        Args:
            hidden_size: 隐藏层维度大小 
            output_size: 输出词汇表大小（目标语言的词汇量）
        """
        super(DecoderRNN, self).__init__()
        # 词嵌入层：将单词索引映射为稠密向量
        self.embedding = nn.Embedding(output_size, hidden_size)  
        # GRU循环神经网络层
        self.gru = nn.GRU(hidden_size, hidden_size)  
        # 输出全连接层：将隐藏状态映射到词汇表空间
        self.out = nn.Linear(hidden_size, output_size)  
        # 对数softmax层：生成概率分布
        self.softmax = nn.LogSoftmax(dim=1)  
        
    def forward(self, input, hidden):
        """
        前向传播过程
        Args:
            input: 当前时间步的输入单词索引（单个词）
            hidden: 上一时间步的隐藏状态
        Returns:
            output: 当前时间步的输出概率分布（log softmax）
            hidden: 当前时间步的隐藏状态
        """
        # 词嵌入+维度调整 (1,1,hidden_size)
        output = self.embedding(input).view(1, 1, -1)  
        # ReLU激活函数
        output = F.relu(output)                     
        # GRU处理 (output: [1,1,hidden_size], hidden: [1,1,hidden_size])
        output, hidden = self.gru(output, hidden)   
        # 全连接层+softmax (output[0]降维到[1,hidden_size])
        output = self.softmax(self.out(output[0]))  
        return output, hidden

    def initHidden(self):
        """初始化全零隐藏状态"""
        return torch.zeros(1, 1, self.hidden_size, device=device)  # [num_layers=1, batch_size=1, hidden_size]
```

```python
class AttenDecoderRNN(nn.Module):
    """带注意力机制的RNN解码器实现"""
    
    def __init__(self, hidden_size, output_size, dropout_p=0.1, max_len=MAX_LENGTH1):
        """
        初始化注意力解码器
        Args:
            hidden_size: 隐藏层维度
            output_size: 输出词汇表大小
            dropout_p: dropout概率（默认0.1）
            max_len: 最大序列长度（用于注意力权重矩阵维度）
        """
        super(AttenDecoderRNN, self).__init__()
        self.hidden_size = hidden_size
        self.output_size = output_size
        self.dropout_p = dropout_p
        self.max_len = max_len

        # 词嵌入层
        self.embedding = nn.Embedding(self.output_size, self.hidden_size)
        
        # 注意力机制相关层
        self.attn = nn.Linear(self.hidden_size * 2, self.max_len)  # 计算注意力权重
        self.attn_combine = nn.Linear(self.hidden_size * 2, self.hidden_size)  # 合并注意力上下文
        
        # 正则化层
        self.dropout = nn.Dropout(self.dropout_p)
        
        # GRU循环层和输出层
        self.gru = nn.GRU(self.hidden_size, self.hidden_size)
        self.out = nn.Linear(self.hidden_size, self.output_size)

    def forward(self, input, hidden, encoder_outputs):
        """
        前向传播过程
        Args:
            input: 当前输入词索引
            hidden: 上一时刻隐藏状态
            encoder_outputs: 编码器所有时间步的输出（用于注意力计算）
        Returns:
            output: 当前输出概率分布（log softmax）
            hidden: 新隐藏状态
            atten_weight: 注意力权重（用于可视化）
        """
        # 1. 词嵌入处理
        embedded = self.embedding(input).view(1, 1, -1)  # [1,1,hidden_size]
        embedded = self.dropout(embedded)  # 应用dropout

        # 2. 计算注意力权重
        atten_weight = F.softmax(
            self.attn(torch.cat([embedded[0], hidden[0]], 1)),  # 拼接当前输入和隐藏状态
            dim=1
        )  # 输出形状: [1, max_len]

        # 3. 应用注意力权重到编码器输出
        att_applied = torch.bmm(
            atten_weight.unsqueeze(0),  # 增加batch维度 -> [1,1,max_len]
            encoder_outputs.unsqueeze(0)  # [1,max_len,hidden_size]
        )  # 输出形状: [1,1,hidden_size]

        # 4. 合并注意力上下文和当前输入
        output = torch.cat([embedded[0], att_applied[0]], dim=1)  # [1, hidden_size*2]
        output = self.attn_combine(output).unsqueeze(0)  # [1,1,hidden_size]
        
        # 5. GRU处理和输出预测
        output = F.relu(output)
        output, hidden = self.gru(output, hidden)
        output = F.log_softmax(self.out(output[0]), dim=1)  # [1, output_size]

        return output, hidden, atten_weight

    def initHidden(self):
        """初始化全零隐藏状态"""
        return torch.zeros(1, 1, self.hidden_size, device=device)  # [1,1,hidden_size]
```

```python
import random
from torch import optim
from utils import timeSince  # 导入计算训练时间的工具函数
import time

MAX_LENGTH = 11  # 定义输入序列的最大长度

# 设置语言和文件路径
lang1 = "en"  # 源语言（英语）
lang2 = "cn"  # 目标语言（中文）
path = r"C:\Users\Lenovo\Downloads\cmn-eng\cmn.txt"  # 数据集路径

# 读取并处理语言数据
input_lang, output_lang, pairs = readLangs(lang1, lang2, path)  # 获取输入/输出语言对象和句子对

def listTotensor(input_lang, data):
    """将句子字符串转换为张量"""
    indexes_in = [input_lang.word2index[word] for word in data.split(" ")]  # 单词转索引
    indexes_in.append(EOS_token)  # 添加句子结束符
    # 创建张量并调整形状为(seq_len, 1)
    input_tensor = torch.tensor(indexes_in,
                                dtype=torch.long,
                                device=device).view(-1, 1)
    return input_tensor

def tensorsFromPair(pair):
    """将句子对转换为输入输出张量"""
    input_tensor = listTotensor(input_lang, pair[0])  # 源语言句子转张量
    output_tensor = listTotensor(output_lang, pair[1])  # 目标语言句子转张量
    return (input_tensor, output_tensor)

def loss_func(input_tensor, output_tensor, encoder, decoder,
             encoder_optimizer, decoder_optimizer,
             scheduler_encoder, scheduler_decoder,
             criterion):
    """计算损失并进行反向传播"""
    encoder_hidden = encoder.initHidden()  # 初始化编码器隐藏状态

    # 清空优化器梯度
    encoder_optimizer.zero_grad()
    decoder_optimizer.zero_grad()

    # 获取输入输出序列长度
    input_len = input_tensor.size(0)
    output_len = output_tensor.size(0)

    # 初始化编码器输出容器
    encoder_outputs = torch.zeros(MAX_LENGTH, encoder.hidden_size, device=device)

    # 编码器前向传播
    for ei in range(input_len):
        encoder_output, encoder_hidden = encoder(input_tensor[ei], encoder_hidden)
        encoder_outputs[ei] = encoder_output[0, 0]  # 存储每个时间步的输出

    # 解码器初始状态使用编码器最终隐藏状态
    decoder_hidden = encoder_hidden
    decoder_input = torch.tensor([[SOS_token]], device=device)  # 解码器起始符
    
    # 随机决定是否使用Teacher Forcing
    use_teacher_forcing = True if random.random() < 0.5 else False

    loss = 0  # 初始化损失
    # Teacher Forcing模式：使用真实标签作为下一输入
    if use_teacher_forcing:
        for di in range(output_len):
            decoder_output, decoder_hidden, decoder_attention = decoder(
                decoder_input, decoder_hidden, encoder_outputs
            )
            # 累积损失（使用负对数似然损失）
            loss += criterion(decoder_output, output_tensor[di])
            decoder_input = output_tensor[di]  # 强制使用真实标签
    # 非Teacher Forcing模式：使用自身预测作为下一输入
    else:
        for di in range(output_len):
            decoder_output, decoder_hidden, decoder_attention = decoder(
                decoder_input, decoder_hidden, encoder_outputs
            )
            loss += criterion(decoder_output, output_tensor[di])
            
            # 获取预测值（概率最高的单词索引）
            topV, topi = decoder_output.topk(1)
            # 分离计算图并调整形状
            decoder_input = topi.squeeze().detach()

            # 若预测为EOS则提前终止
            if decoder_input.item() == EOS_token:
                break

    # 反向传播
    loss.backward()          
    # 参数更新
    encoder_optimizer.step()  
    decoder_optimizer.step()  
    return loss.item() / output_len  # 返回平均损失

# 初始化模型参数
hidden_size = 256  # 隐藏层维度
encoder = EncoderRNN(input_lang.n_words, hidden_size).to(device)  # 编码器
# 带注意力机制的解码器
decoder = AttenDecoderRNN(hidden_size,
                          output_lang.n_words,
                          max_len = MAX_LENGTH,
                          dropout_p = 0.1).to(device)

# 优化器配置
lr = 0.01  # 初始学习率
encoder_optimizer = optim.SGD(encoder.parameters(), lr=lr)  # 编码器优化器
decoder_optimizer = optim.SGD(decoder.parameters(), lr=lr)  # 解码器优化器

# 学习率调度器（每1个epoch衰减5%）
scheduler_encoder = torch.optim.lr_scheduler.StepLR(encoder_optimizer,
                                                    step_size=1,
                                                    gamma=0.95)
scheduler_decoder = torch.optim.lr_scheduler.StepLR(decoder_optimizer,
                                                    step_size=1,
                                                    gamma=0.95)

criterion = nn.NLLLoss()  # 负对数似然损失函数

# 训练配置
n_iters = 20000  # 总迭代次数
# 随机选择训练样本对
training_pairs = [tensorsFromPair(random.choice(pairs)) for i in range(n_iters)]

print_every = 200  # 打印间隔
save_every = 2000   # 模型保存间隔

print_loss_total = 0  # 累计损失
start = time.time()   # 训练开始时间

# 训练循环
for iter in range(1, n_iters + 1):
    training_pair = training_pairs[iter - 1]  # 获取当前训练对
    input_tensor = training_pair[0]  # 输入张量
    output_tensor = training_pair[1]  # 输出张量

    # 计算损失并更新模型
    loss = loss_func(input_tensor,
                     output_tensor,
                     encoder,
                     decoder,
                     encoder_optimizer,
                     decoder_optimizer,
                     scheduler_encoder,
                     scheduler_decoder,
                     criterion)
    print_loss_total += loss  # 累计损失

    # 定期打印训练信息
    if iter % print_every == 0:
        print_loss_avg = print_loss_total / print_every  # 计算平均损失
        print_loss_total = 0  # 重置累计损失
        # 打印进度信息：耗时/迭代数/进度百分比/平均损失
        print("{},{},{},{}".format(timeSince(start, iter/n_iters),
                                   iter, iter / n_iters * 100,
                                   print_loss_avg))

    # 定期保存模型
    if iter % save_every == 0:
        torch.save(encoder.state_dict(),
                   "models/encoder_{}.pth".format(iter))
        torch.save(decoder.state_dict(),
                   "models/decoder_{}.pth".format(iter))

    # 每10000次迭代更新学习率
    if iter % 10000 == 0:
        scheduler_encoder.step()  # 更新编码器学习率
        scheduler_decoder.step()  # 更新解码器学习率
```
