#不定长参数,用于函数定义参数个数不确定的场景


```python
def calc_data(*arges):#位置传递的不定长参数
    min_data=min(arges)
    max_data=max(arges)
    avg_data=sum(arges)/len(arges)
    return min_data,max_data,avg_data
```

```python
print(calc_data(12,25,62,156,515,33,18,15))
```

```python
#关键字传递的不定长参数,关键字传递封装以后他是字典的形式
def calc_data(*arges,**kwarges):
    min_data=min(arges)
    max_data=max(arges)
    avg_data=sum(arges)/len(arges)
    
    return min_data,max_data,round(avg_data,1)
```
