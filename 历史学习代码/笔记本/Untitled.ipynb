```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re

# 数据导入
df = pd.read_excel('./中国电影网电影_data.xlsx')

# 清洗票房列
def clean_box_office(value):
    if pd.isna(value):
        return np.nan
    try:
        return float(value)
    except:
        pass
    if isinstance(value, str):
        cleaned = re.sub(r'[^\d.]', '', value)
        if cleaned:
            return float(cleaned)
        if '亿' in value:
            num_part = re.search(r'[\d.]+', value)
            if num_part:
                return float(num_part.group()) * 10000
        if '万' in value:
            num_part = re.search(r'[\d.]+', value)
            if num_part:
                return float(num_part.group())
    return np.nan

df['累计票房'] = df['累计票房'].apply(clean_box_office)

# 处理日期
df['上映时间'] = df['上映时间'].str.replace(r'[（）\(\)].*', '', regex=True)
df['上映时间'] = pd.to_datetime(df['上映时间'], errors='coerce')
df['dy_year'] = df['上映时间'].dt.year

# 分组统计
yearly_data = df.groupby('dy_year').agg(
    movie_count=('电影名', 'count'),
    total_box_office=('累计票房', 'sum')
).reset_index()

# 绘制图表
fig, ax1 = plt.subplots(figsize=(12, 6))
ax1.plot(yearly_data['dy_year'], yearly_data['movie_count'], 
         'b-o', linewidth=2, label='电影数量')
ax1.set_xlabel('年份')
ax1.set_ylabel('电影数量', color='b')
ax1.tick_params(axis='y', labelcolor='b')

ax2 = ax1.twinx()
ax2.bar(yearly_data['dy_year'], yearly_data['total_box_office']/10000,
        color='r', alpha=0.6, label='总票房(亿)')
ax2.set_ylabel('总票房(亿元)', color='r')
ax2.tick_params(axis='y', labelcolor='r')

plt.title('中国电影年度趋势 (2010-2023)', fontsize=14)
plt.grid(ls='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
df['上映日期'] = pd.to_datetime(df['上映日期'], errors='coerce')  # 添加errors='coerce'处理异常值
df['dy_year'] = df['上映日期'].dt.year

# 检查是否有无效日期
if df['上映日期'].isnull().any():
    print(f"警告: 有 {df['上映日期'].isnull().sum()} 条记录的日期转换失败")

# 年代统计分析
year_stats = df['dy_year'].describe()
print("\n年代统计分析:")
print(year_stats)
```
