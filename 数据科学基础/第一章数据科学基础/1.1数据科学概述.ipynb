```python
prices = [12, 8, 5]

def total_price(values):
    total = 0
    for value in values:
        if value >= 0:
            total += value
    return total

print(total_price(prices))
```

#计算角度转换为弧度

```python
import math
angle=float(input('请输入角度:'))
radian=math.pi/180*angle
Sinf=math.sin(radian)
print('弧度',radian)
print('sin的函数值',Sinf)
```
