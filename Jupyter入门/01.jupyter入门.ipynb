```python
score_df=pd.DataFrame(np.random.randint(0,101,(10,5)))
print(score_df)
```

```python
#修改列名和索引值
column_name=['语文','数学','英语','体育','政治']
index_name=['同学'+str(i) for i in range(10)]
score_df.columns=column_name
score_df.index=index_name
print(score_df)
```

```python
#Dataframe的属性
print(score_df.shape)
print(score_df.columns)
print(score_df.index)
print(score_df.T)
```
