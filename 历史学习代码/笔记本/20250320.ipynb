```python
label = []

for word in word_list:
    if len(word)==1:
        label.append('S')
    else:
        label += ['B'] + (len(word)-2)*['M'] + ['E']
```

初始化概率init_pro

```python
init_pro = {}

for state in state_list:
    init_pro[state] = 0
```

```python
init_pro
```

转移概率trans_pro

```python
trans_pro = {}

for state in state_list: #循环
    
    trans = {}
    for s in state_list: #嵌套循环
        trans[s] = 0
    
    trans_pro[state] = trans
```
