```python
state_list = ['B','M','E','S']
init_prob = {}
for state in state_list:
    init_prob[state]=0
trans_prob = {}
for state in state_list:
    trans = {}
    for sta in state_list:
        trans[sta] = 0
    trans_prob[state]=trans
emit_prob = {}
for state in state_list:
    emit_prob[state] = collections.defaultdict(int)
```

```python
for line in lines:
    line = line.strip()
    word_list = line.split()
    label_list = []
    for word in word_list:
        if len(word)==1:
            label_list.append('S')
        else:
            label_list += ['B']  + (len(word)-2)*['M']  +  ['E']
        
    for index in range(len(label_list)):
        if index==0:
            init_prob[label_list[index]]+=1
        else:
            trans_prob[label_list[index-1]][label_list[index]] += 1
            
        w_line=line.replace(' ','')
        w_list=list(w_line)
    for index in range(len(label_list)):
        emit_prob[label_list[index]][w_list[index]]+=1
        
        
```
