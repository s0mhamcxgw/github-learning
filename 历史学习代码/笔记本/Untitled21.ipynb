```python
# 导入必要的库
from pyecharts.charts import Bar
from pyecharts import options as opts
from pyecharts.globals import ThemeType
from pyecharts.commons.utils import JsCode  # 用于自定义渐变

# 定义数据集
goods = ["衬衫", "羊毛衫", "雪纺衫", "裤子", "高跟鞋", "袜子"]
sales1 = [5, 20, 36, 10, 10, 100]  # 商家A销售量
sales2 = [55, 60, 16, 20, 15, 80]  # 商家B销售量

# 1. 商家A单柱形图（带渐变效果）
bar1 = (
    Bar(init_opts=opts.InitOpts(theme=ThemeType.LIGHT, width="800px", height="500px"))
    .add_xaxis(goods)
    .add_yaxis(
        "商家A", 
        sales1,
        # 添加标签显示具体数值
        label_opts=opts.LabelOpts(position="top", font_size=12),
        # 添加渐变色效果 - 使用JsCode替代LinearGradientOpts
        itemstyle_opts=opts.ItemStyleOpts(
            color=JsCode("""
                new echarts.graphic.LinearGradient(0, 0, 0, 1, [{
                    offset: 0,
                    color: '#5470C6'
                }, {
                    offset: 1,
                    color: '#91CC75'
                }])
            """)
        )
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="商家A商品销售量分析", subtitle="单位：件"),
        yaxis_opts=opts.AxisOpts(
            name="销售量",
            name_location="end",
            name_gap=20,
            min_=0,
            max_=110
        ),
        # 添加工具箱（可下载图片）
        toolbox_opts=opts.ToolboxOpts(is_show=True)
    )
)
bar1.render_notebook()
bar1.render("商家A单柱形图.html")

# 2. 双商家柱形图（带标记点和标记线）
bar2 = (
    Bar(init_opts=opts.InitOpts(theme=ThemeType.MACARONS, width="900px", height="550px"))
    .add_xaxis(goods)
    .add_yaxis(
        "商家A", 
        sales1,
        gap="40%",  # 设置柱间距
        # 添加平均值标记点
        markpoint_opts=opts.MarkPointOpts(
            data=[opts.MarkPointItem(type_="average", name="平均值")]
        ),
        # 自定义柱形颜色
        itemstyle_opts=opts.ItemStyleOpts(color="#EE6666")
    )
    .add_yaxis(
        "商家B", 
        sales2,
        # 添加最大最小值标记线
        markline_opts=opts.MarkLineOpts(
            data=[
                opts.MarkLineItem(type_="min", name="最小值"),
                opts.MarkLineItem(type_="max", name="最大值")
            ],
            linestyle_opts=opts.LineStyleOpts(width=3, type_="dashed")
        ),
        # 自定义柱形颜色
        itemstyle_opts=opts.ItemStyleOpts(color="#91CC75")
    )
    .set_global_opts(
        title_opts=opts.TitleOpts(title="商家A/B销售对比分析", subtitle="单位：件"),
        yaxis_opts=opts.AxisOpts(name="销售量", min_=0, max_=120),
        # 添加数据缩放功能
        datazoom_opts=opts.DataZoomOpts(type_="inside"),
        # 添加图例配置
        legend_opts=opts.LegendOpts(pos_left="right", orient="vertical")
    )
    .set_series_opts(
        # 添加数值标签
        label_opts=opts.LabelOpts(is_show=True, position="top")
    )
)
bar2.render_notebook()
bar2.render("双商家柱形图.html")

# 3. 水平条形图（双商家对比）
bar3 = (
    Bar(init_opts=opts.InitOpts(theme=ThemeType.ESSOS, width="850px", height="600px"))
    .add_xaxis(goods)
    .add_yaxis("商家A", sales1, 
               category_gap="30%",
               itemstyle_opts=opts.ItemStyleOpts(color="#5470C6"))
    .add_yaxis("商家B", sales2, 
               category_gap="30%",
               itemstyle_opts=opts.ItemStyleOpts(color="#91CC75"))
    # 转换为水平条形图
    .reversal_axis()
    .set_global_opts(
        title_opts=opts.TitleOpts(title="商家A/B销售对比（水平视图）", subtitle="单位：件"),
        xaxis_opts=opts.AxisOpts(name="销售量", min_=0, max_=120),
        # 添加图例配置
        legend_opts=opts.LegendOpts(pos_top="bottom", pos_left="center"),
        # 添加视觉映射组件
        visualmap_opts=opts.VisualMapOpts(
            orient="horizontal",
            pos_left="center",
            min_=0,
            max_=100,
            range_text=["高销量", "低销量"],
            dimension=0,
            range_color=["#D7DA8B", "#E15457"]
        )
    )
    .set_series_opts(
        # 添加数值标签
        label_opts=opts.LabelOpts(position="right"),
        # 添加柱形样式
        itemstyle_opts=opts.ItemStyleOpts(
            border_radius=5,  # 圆角效果
            opacity=0.8  # 透明度
        )
    )
)
bar3.render_notebook()
bar3.render("水平条形图.html")
```
