```python
n=int(input("请输入你要输出的行数"))
list=[]
for i in range(n):
    now=[1]*(i+1)
    for j in range(1,i):
        now[j]=list[i-1][j-1]+list[i-1][j]
    list.append(now)
for line in list:
    print(line)
```

```python
def fin(n):
    if n==0 or n==1:
        return 1
    else:
        return n*fin(n-1)

n=int(input("请输入你要求的阶乘数："))
result=fin(n)
print(f"所求的阶乘结果是{result}")
```
