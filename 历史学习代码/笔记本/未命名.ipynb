```python
#冒泡
lst = [5, 2, 9, 1, 3]
for i in range(len(lst)):
    for j in range(len(lst) - i - 1):
        if lst[j]>lst[j+1]:
            lst[j],lst[j+1]=lst[j+1],lst[j]
print(lst)
```

```python
#简单选择
lst = [5, 2, 9, 1, 3]
for i in range(len(lst)):
    min_index=i
    for j in range(i+1,len(lst)):
        if lst[j]<lst[min_index]:
            min_index=j
    lst[i],lst[min_index]=lst[min_index],lst[i]
print(lst)
    
```

```python
# 二分插入排序（最简单版本）
def binary_sort(lst):
    # 从第2个元素开始，逐个插入
    for i in range(1, len(lst)):
        key = lst[i]  # 要插入的数
        left = 0
        right = i - 1

        # 二分查找：找插入位置
        while left <= right:
            mid = (left + right) // 2  # 中间下标
            if key < lst[mid]:
                right = mid - 1
            else:
                left = mid + 1

        # 把元素后移，空出位置
        for j in range(i, left, -1):
            lst[j] = lst[j - 1]

        # 插入到正确位置
        lst[left] = key

# 测试
lst = [5, 2, 9, 1, 3]
binary_sort(lst)
print(lst)
```
