```python
import pandas as pd
```

```python
price=pd.DataFrame({'frult':['apple','grape','orange','orange'],'price':[8,7,9,11]})
```

```python
amount=pd.DataFrame({'frult':['apple','grape','orange'],'amount':[5,11,8]})
```

```python
display(price,amount,pd.merge(price,amount))
```

```python
display(pd.merge(price,amount,left_on='frult',right_on='frult'))
```

```python
display(pd.merge(price,amount,how='left'))
```

```python
display(pd.merge(price,amount,how='right'))
```

```python
left=pd.DataFrame({'key1':['one','one','two'],'key2':['a','b','a'],'value':range(3)})
```

```python
right=pd.DataFrame({'key1':['one','one','two','two'],'key2':['a','b','a','b'],'value2':range(4)})
```

```python
display(left,right,pd.merge(left,right,on=['key1','key2'],how = 'left'))
```

```python
print(pd.merge(left,right,on='key1'))
```

```python
print(pd.merge(left,right,on='key1',suffixes=('_left','_right')))
```

```python
s1=pd.Series([0,1],index=["a","b"])
s2=pd.Series([2,3,4],index=["a","d",'e'])
s3=pd.Series([5,6],index=["f","g"])
s4=pd.concat([s1*5,s3],sort=False)
s5=pd.concat([s1,s4],axis=1,sort=False)
s6=pd.concat([s1,s4],axis=1,join='inner',sort=False)
display(s5,s6)
```

```python
# 例3-32 使用combine_first()函数合并w1和w2
import pandas as pd
from IPython.display import display

w1 = pd.DataFrame({0: [0, 1], 1: [0, 5]}, index=['a', 'b'])
w2 = pd.DataFrame({0: [0.0, 1.0, float('nan'), float('nan')],
                   1: [0, 5, 5, 6]}, index=['a', 'b', 'f', 'g'])
display(w1, w2, w1.combine_first(w2))
```
